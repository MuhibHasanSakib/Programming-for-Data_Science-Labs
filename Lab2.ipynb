Course Marks with a list

```python
programming_ds= float(input("Enter your Programming for DS mark:"))
data_visualization=float(input("Enter your Data Visualization mark:"))
data_processing=float(input("Enter your Data Processing mark:"))
python=float(input("Enter your Python mark:"))
java= float(input("Enter your Java mark:"))
marks=[programming_ds,data_visualization, data_processing, python, java]
total_marks=0

for mark in marks:
  total_marks+=mark
  average_grade=total_marks/len(marks)

highest_grade= marks[0]
lowest_grade=marks[0]
for mark in marks:
  if mark > highest_grade:
    highest_grade=mark
  if mark < lowest_grade:
    lowest_grade=mark
count=0;
for mark in marks:
  if mark>=50:
    count+=1
print("Total marks=",total_marks)
print("Average_grade", average_grade)
print("Highest Grade:",highest_grade)
print("Lowest grade",lowest_grade)
print("Passed courses:", count)
```

Shopping cart processor

```python
cart = [
    {"name": "Laptop", "price": 1000, "quantity": 3},
    {"name": "Mouse", "price": 20, "quantity": 3},
    {"name": "Keyboard", "price": 50, "quantity": 3},
    {"name": "Monitor", "price": 300, "quantity": 1},
]
total = 0
for item in cart:
    total += item["price"] * item["quantity"]
print("Total cost", total)
def apply_discount(total):
  if total >= 2000:
        return total * 0.90
  return total
final_price=apply_discount(total)
print("Discounted price", final_price)
```

```python
students = [
    {"name": "Rahim", "department": "CSE",
     "marks": 85, "attendance": 92},
    {"name": "Karim", "department": "CSE",
     "marks": 48, "attendance": 76},
    {"name": "Sara", "department": "DS",
     "marks": 91, "attendance": 95},
    {"name": "Nadia", "department": "DS",
     "marks": 67, "attendance": 84},
    {"name": "Hasan", "department": "CSE",
     "marks": 73, "attendance": 78}
]
total = 0
passed = 0
top_name = ""
top_marks = 0

for x in students:
    print(x["name"], "-", x["marks"])
    total = total + x["marks"]

    if x["marks"] >= 50:
        passed = passed + 1

    if x["marks"] > top_marks:
        top_marks = x["marks"]
        top_name = x["name"]

average = total / len(students)

print("Students passed:", passed)
print("Class average:", average)
print("Highest mark:", top_name, "with", top_marks)
```

```python
products = [
    {"name": "Notebook", "price": 120, "quantity": 3},
    {"name": "Pen", "price": 20, "quantity": 5},
    {"name": "USB Drive", "price": 850, "quantity": 2},
    {"name": "Mouse", "price": 650, "quantity": 1}
]

inventory_total = 0
top_name = ""
top_line_total = 0

for p in products:
    line_total = p["price"] * p["quantity"]
    print(p["name"], "line total =", line_total)
    inventory_total = inventory_total + line_total

    if line_total > top_line_total:
        top_line_total = line_total
        top_name = p["name"]

print("Total inventory value:", inventory_total)
print("Highest line total:", top_name, "with", top_line_total)
```
