Course Marks with a list

```python
programming_ds= float(input("Enter your Programming for DS mark:"))
data_visualization=float(input("Enter your Data Visualization mark:"))
data_processing=float(input("Enter your Data Processing mark:"))
python=float(input("Enter your Python mark:"))
java= float(input("Enter your Java mark:"))
marks=[programming_ds,data_visualization, data_processing, python, java]
total_marks=0

for mark in marks:
  total_marks+=mark
  average_grade=total_marks/len(marks)

highest_grade= marks[0]
lowest_grade=marks[0]
for mark in marks:
  if mark > highest_grade:
    highest_grade=mark
  if mark < lowest_grade:
    lowest_grade=mark
count=0;
for mark in marks:
  if mark>=50:
    count+=1
print("Total marks=",total_marks)
print("Average_grade", average_grade)
print("Highest Grade:",highest_grade)
print("Lowest grade",lowest_grade)
print("Passed courses:", count)
```

Shopping cart processor

```python
cart = [
    {"name": "Laptop", "price": 1000, "quantity": 3},
    {"name": "Mouse", "price": 20, "quantity": 3},
    {"name": "Keyboard", "price": 50, "quantity": 3},
    {"name": "Monitor", "price": 300, "quantity": 1},
]
total = 0
for item in cart:
    total += item["price"] * item["quantity"]
print("Total cost", total)
def apply_discount(total):
  if total >= 2000:
        return total * 0.90
  return total
final_price=apply_discount(total)
print("Discounted price", final_price)
```
